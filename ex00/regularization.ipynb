# Regularization

## 0. Imports

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import accuracy_score
```

## 1. Preprocessing

```python
df = pd.read_csv('../data/dayofweek.csv')
X = df.drop('dayofweek', axis=1).values
y = df['dayofweek'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=21, stratify=y
)
print('Train shape:', X_train.shape)
print('Test shape:', X_test.shape)
```

```python
def crossval_eval(model, X_train, y_train, n_splits=10):
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=21)
    val_scores = []
    for train_idx, val_idx in skf.split(X_train, y_train):
        X_tr, X_val = X_train[train_idx], X_train[val_idx]
        y_tr, y_val = y_train[train_idx], y_train[val_idx]
        model.fit(X_tr, y_tr)
        train_acc = accuracy_score(y_tr, model.predict(X_tr))
        val_acc = accuracy_score(y_val, model.predict(X_val))
        print(f'train -  {train_acc:.5f}   |   valid -  {val_acc:.5f}')
        val_scores.append(val_acc)
    print(f'Average accuracy on crossval is {np.mean(val_scores):.5f}')
    print(f'Std is {np.std(val_scores):.5f}')
    return np.mean(val_scores)
```

## 2. Logreg regularization

### a. Default regularization

```python
%%time
logreg = LogisticRegression(random_state=21, fit_intercept=False, max_iter=1000)
crossval_eval(logreg, X_train, y_train)
```

### b. Optimizing regularization parameters

```python
# penalty=None
logreg_none = LogisticRegression(penalty=None, random_state=21, fit_intercept=False, max_iter=1000)
print('penalty=None:')
crossval_eval(logreg_none, X_train, y_train)
```

```python
# penalty=l1
logreg_l1 = LogisticRegression(penalty='l1', solver='saga', random_state=21, fit_intercept=False, max_iter=1000)
print('penalty=l1:')
crossval_eval(logreg_l1, X_train, y_train)
```

```python
# penalty=l2
logreg_l2 = LogisticRegression(penalty='l2', random_state=21, fit_intercept=False, max_iter=1000)
print('penalty=l2:')
crossval_eval(logreg_l2, X_train, y_train)
```

## 3. SVM regularization

### a. Default regularization

```python
%%time
svm = SVC(probability=True, kernel='linear', random_state=21)
crossval_eval(svm, X_train, y_train)
```

### b. Optimizing regularization parameters

```python
for C in [0.01, 0.1, 1, 5, 10]:
    svm_c = SVC(C=C, probability=True, kernel='linear', random_state=21)
    scores = []
    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=21)
    for tr_idx, val_idx in skf.split(X_train, y_train):
        svm_c.fit(X_train[tr_idx], y_train[tr_idx])
        scores.append(accuracy_score(y_train[val_idx], svm_c.predict(X_train[val_idx])))
    print(f'C={C}: mean={np.mean(scores):.5f}')
```

```python
svm_best = SVC(C=10, probability=True, kernel='rbf', random_state=21)
print('SVM rbf C=10:')
crossval_eval(svm_best, X_train, y_train)
```

```python
svm_balanced = SVC(C=1, probability=True, kernel='rbf', class_weight='balanced', random_state=21)
print('SVM rbf balanced:')
crossval_eval(svm_balanced, X_train, y_train)
```

## 4. Tree

### a. Default regularization

```python
%%time
tree = DecisionTreeClassifier(max_depth=10, random_state=21)
crossval_eval(tree, X_train, y_train)
```

### b. Optimizing regularization parameters

```python
for depth in [5, 10, 15, 20, 25, 30]:
    t = DecisionTreeClassifier(max_depth=depth, random_state=21)
    scores = []
    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=21)
    for tr_idx, val_idx in skf.split(X_train, y_train):
        t.fit(X_train[tr_idx], y_train[tr_idx])
        scores.append(accuracy_score(y_train[val_idx], t.predict(X_train[val_idx])))
    print(f'max_depth={depth}: mean={np.mean(scores):.5f}')
```

```python
tree_best = DecisionTreeClassifier(max_depth=20, min_samples_leaf=2, random_state=21)
print('Tree max_depth=20, min_samples_leaf=2:')
crossval_eval(tree_best, X_train, y_train)
```

```python
tree_b2 = DecisionTreeClassifier(max_depth=20, min_samples_split=5, random_state=21)
print('Tree max_depth=20, min_samples_split=5:')
crossval_eval(tree_b2, X_train, y_train)
```

```python
tree_b3 = DecisionTreeClassifier(max_depth=20, max_leaf_nodes=100, random_state=21)
print('Tree max_depth=20, max_leaf_nodes=100:')
crossval_eval(tree_b3, X_train, y_train)
```

## 5. Random forest

### a. Default regularization

```python
%%time
rf = RandomForestClassifier(n_estimators=50, max_depth=14, random_state=21)
crossval_eval(rf, X_train, y_train)
```

### b. Optimizing regularization parameters

```python
for depth in [10, 14, 18, 22]:
    for n_est in [50, 100, 200]:
        r = RandomForestClassifier(n_estimators=n_est, max_depth=depth, random_state=21)
        scores = []
        skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=21)
        for tr_idx, val_idx in skf.split(X_train, y_train):
            r.fit(X_train[tr_idx], y_train[tr_idx])
            scores.append(accuracy_score(y_train[val_idx], r.predict(X_train[val_idx])))
        print(f'max_depth={depth}, n_estimators={n_est}: mean={np.mean(scores):.5f}')
```

```python
rf_best = RandomForestClassifier(n_estimators=200, max_depth=22, min_samples_leaf=2, random_state=21)
print('RF n_estimators=200, max_depth=22, min_samples_leaf=2:')
crossval_eval(rf_best, X_train, y_train)
```

```python
rf_b2 = RandomForestClassifier(n_estimators=100, max_depth=18, max_features='sqrt', random_state=21)
print('RF max_features=sqrt:')
crossval_eval(rf_b2, X_train, y_train)
```

```python
rf_b3 = RandomForestClassifier(n_estimators=200, max_depth=22, class_weight='balanced', random_state=21)
print('RF balanced:')
crossval_eval(rf_b3, X_train, y_train)
```

## 6. Predictions

```python
# Best model is Random Forest
best_model = RandomForestClassifier(n_estimators=200, max_depth=22, random_state=21)
best_model.fit(X_train, y_train)
y_pred = best_model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f'Final accuracy on test: {acc:.5f}')
```

```python
# Error analysis per weekday
errors = pd.DataFrame({'true': y_test, 'pred': y_pred})
errors['wrong'] = errors['true'] != errors['pred']
error_rate = errors.groupby('true')['wrong'].mean().sort_values(ascending=False)
print('Error rate per weekday (0=Mon, 6=Sun):')
print(error_rate)
```

```python
# Save model
import pickle
with open('../data/regularization_model.pkl', 'wb') as f:
    pickle.dump(best_model, f)
print('Model saved!')
```
